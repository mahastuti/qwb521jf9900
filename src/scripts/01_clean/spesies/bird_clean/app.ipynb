```python
def descriptive_stats(data, name):
    print(f"\nDescriptive Statistics - {name}")
    df = pd.Series(data)
    print(df.describe())

descriptive_stats(day1, "Day 1")
descriptive_stats(day2, "Day 2")
descriptive_stats(comp1, "Company 1")
descriptive_stats(comp2, "Company 2")
```

```python
plt.figure(figsize=(8,5))
plt.boxplot([day1, day2], labels=["Day 1", "Day 2"])
plt.axhline(2270, color='red', linestyle='--', label="Nominal 2.27 L")
plt.legend()
plt.title("Boxplot Data Day 1 & Day 2")
plt.show()

plt.figure(figsize=(8,5))
plt.boxplot([comp1, comp2], labels=["Company 1", "Company 2"])
plt.axhline(2270, color='red', linestyle='--')
plt.title("Boxplot Company 1 & Company 2")
plt.show()
```
