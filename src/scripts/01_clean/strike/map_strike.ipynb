```python
import pandas as pd

# =========================
# CONFIG: ganti path file
# =========================
DATA1_PATH = "data1.csv"      # file utama (download dari GSheets)
FOREIGN_PATH = "foreign.csv"  # file tabel foreign (mapping)
OUT_PATH = "data1_supabase.csv"

# =========================
# Helper: baca file dengan delimiter yang "ketebak"
# =========================
def read_table(path: str) -> pd.DataFrame:
    # Coba tab dulu (sering dari copy/export tertentu), kalau gagal baru koma/semicolon
    for sep in ["\t", ",", ";", "|"]:
        try:
            df = pd.read_csv(path, sep=sep, dtype=str, engine="python")
            # anggap valid kalau kolomnya > 1
            if df.shape[1] > 1:
                return df
        except Exception:
            pass
    raise ValueError(f"Gagal membaca file {path}. Coba cek delimiter/formatnya.")

def norm_series(s: pd.Series) -> pd.Series:
    return (
        s.fillna("")
         .astype(str)
         .str.replace("\ufeff", "", regex=False)   # buang BOM kalau ada
         .str.strip()
         .str.replace(r"\s+", " ", regex=True)     # rapikan spasi/newline
         .str.lower()
    )

# =========================
# Load data
# =========================
df = read_table(DATA1_PATH)
fk = read_table(FOREIGN_PATH)

# =========================
# Pastikan kolom yang dibutuhkan ada
# =========================
required_fk_cols = [
    "phase_id", "phase",
    "perimeter_id", "perimeter",
    "category_id", "category",
    "remark_id", "remark",
    "airline_id", "airline",
    "component_id", "component",
    "type_id", "type",
]
missing_fk = [c for c in required_fk_cols if c not in fk.columns]
if missing_fk:
    raise ValueError(f"Kolom foreign missing: {missing_fk}")

# Kolom id di data1 yang MAU dikonversi (kecuali point_id)
id_cols_to_convert = ["phase_id", "perimeter_id", "category_id", "remark_id", "airline_id", "component_id", "type_id"]

missing_df = [c for c in id_cols_to_convert if c not in df.columns]
if missing_df:
    raise ValueError(f"Kolom data1 missing: {missing_df}")

# =========================
# Normalisasi teks untuk matching
# =========================
# Normalisasi kolom nama pada foreign
for col in ["phase", "perimeter", "category", "remark", "airline", "component", "type"]:
    fk[col] = norm_series(fk[col])

# Normalisasi nilai pada data1 (yang saat ini masih berupa nama, tapi disimpan di kolom *_id)
for col in id_cols_to_convert:
    df[col] = norm_series(df[col])

# =========================
# Buat mapping: nama -> id
# =========================
maps = {
    "phase_id": dict(zip(fk["phase"], fk["phase_id"].astype(str))),
    "perimeter_id": dict(zip(fk["perimeter"], fk["perimeter_id"].astype(str))),
    "category_id": dict(zip(fk["category"], fk["category_id"].astype(str))),
    "remark_id": dict(zip(fk["remark"], fk["remark_id"].astype(str))),
    "airline_id": dict(zip(fk["airline"], fk["airline_id"].astype(str))),
    "component_id": dict(zip(fk["component"], fk["component_id"].astype(str))),
    "type_id": dict(zip(fk["type"], fk["type_id"].astype(str))),
}

# =========================
# Konversi kolom di data1: nama -> id
# =========================
for idcol in id_cols_to_convert:
    df[idcol] = df[idcol].map(maps[idcol])

# =========================
# Report unmatched (biar ketahuan yang belum ada di foreign)
# =========================
print("=== UNMATCHED REPORT ===")
for idcol in id_cols_to_convert:
    n_bad = df[idcol].isna().sum()
    print(f"{idcol}: {n_bad} baris belum ke-map")
    if n_bad > 0:
        # tampilkan contoh 10 nilai asli (sebelum map) tidak ada lagi,
        # jadi kita baca ulang kolom mentah dari file untuk contoh:
        pass

# Kalau kamu mau drop baris yang belum ke-map (opsional):
# df = df.dropna(subset=id_cols_to_convert)

# =========================
# Export CSV delimiter koma untuk Supabase
# =========================
df.to_csv(OUT_PATH, index=False, encoding="utf-8")
print(f"\nDONE -> {OUT_PATH}")
```

```python
import pandas as pd
import numpy as np

# 1. Load data
df = pd.read_csv("birdstrike.csv")

# 2. Daftar kolom FK / ID
id_cols = [
    "event_id",
    "phase_id",
    "perimeter_id",
    "category_id",
    "remark_id",
    "airline_id",
    "component_id",
    "type_id",
    "point_id",
]

# 3. Bersihin kolom ID
for col in id_cols:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)              # amankan dari mixed type
            .str.strip()              # TRIM spasi
            .replace({"nan": np.nan, "": np.nan})
        )
        df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

# 4. (Opsional tapi disarankan) buang baris kosong total
df = df.dropna(how="all")

# 5. Save ulang ke CSV bersih
df.to_csv("birdstrike_clean.csv", index=False)

print("✅ Cleaning selesai. File: birdstrike_clean.csv")
```

```python
import pandas as pd
import numpy as np

# Load CSV bersih kamu
df = pd.read_csv("birdstrike_final.csv")

# Pastikan event_id gak ikut (kalau masih ada)
df = df.drop(columns=["event_id"], errors="ignore")

# Kolom bigint kamu
bigint_cols = ["phase_id","perimeter_id","category_id","remark_id",
               "airline_id","component_id","type_id","point_id"]

# Normalisasi: kosong/whitespace -> NaN -> NULL
for c in bigint_cols:
    if c in df.columns:
        df[c] = (
            df[c].astype(str).str.strip()
            .replace({"nan": np.nan, "None": np.nan, "": np.nan})
        )
        df[c] = pd.to_numeric(df[c], errors="coerce").astype("Int64")

# Helper escape string SQL
def sql_str(x):
    if pd.isna(x):
        return "NULL"
    s = str(x)
    s = s.replace("'", "''")   # escape single quote
    return f"'{s}'"

def sql_int(x):
    if pd.isna(x):
        return "NULL"
    return str(int(x))

cols = list(df.columns)

values_sql = []
for _, row in df.iterrows():
    vals = []
    for c in cols:
        if c in bigint_cols:
            vals.append(sql_int(row[c]))
        else:
            vals.append(sql_str(row[c]))
    values_sql.append("(" + ",".join(vals) + ")")

sql = f"INSERT INTO public.bird_strike ({', '.join(cols)})\nVALUES\n" + ",\n".join(values_sql) + ";\n"

with open("birdstrike_insert.sql", "w", encoding="utf-8") as f:
    f.write(sql)

print("✅ Generated: birdstrike_insert.sql")
```
